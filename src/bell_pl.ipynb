```python
import pennylane as qml
```

```python
dev = qml.device('default.qubit', wires = 2);
```

```python
@qml.qnode(dev)
def bell():
    qml.Hadamard(wires = 0);
    qml.CNOT(wires = [0, 1]);
    return qml.state()
```

```python
res = bell()
res
```

```python
print(qml.draw(bell)())
```
