```python
from qiskit import *
from qiskit_aer import *
from qiskit.quantum_info import Statevector, SparsePauliOp
```

```python
def circuit(n, state, layers):
    assert len(state) == n;
    
    qc = QuantumCircuit(n);
    for idx, ch in enumerate(state):
        if ch == '1':
            qc.x(idx);

    for layer in layers:
        assert len(layer) == n;
        for idx, qubit_weights in enumerate(layer):
            assert len(qubit_weights) == 3;
            qc.u(*qubit_weights, idx)
    
    return qc;
```

```python
n = 5
test = [[[0] * 3] * n] * 2
state = '10110'
qc = circuit(n, state, test);
```
