```python
import pennylane as qml
from pennylane import numpy as np
```

```python
def init_graph():
    n = int(input("Enter the number of vertices in the graph: "));
    e = int(input("Enter the number of edges in the graph: "));

    adj = list();
    for _ in range(e):
        a, b = [int(x) for x in input().split()]
        adj.append(list([a, b]));

    return [n, adj]
```

```python
def max_cut_ham(adj: list[list[int]]):
    H = [[], []]

    for edge in adj:
        a = edge[0]; a -= 1;
        b = edge[1]; b -= 1;
        H += 0.5 * qml.Identity(a) - 0.5 * (qml.PauliZ(a) @ qml.PauliZ(b));

    return H
```

```python
num_qubits, adj = init_graph()
h = max_cut_ham(adj)
```

```python
dev = qml.device('default.qubit', wires = num_qubits)
```

```python
@qml.qnode(dev)
def ckt(params):
    for i in range(num_qubits):
        qml.RX(params[i], wires = i)
    return qml.expval(h)
```
