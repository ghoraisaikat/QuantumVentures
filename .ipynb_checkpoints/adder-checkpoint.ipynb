```python
from qiskit import *
from qiskit_aer import *
```

```python
import numpy as np
```

```python
maj = QuantumCircuit(3, name = "MAJ");
maj.cx(2, 1); maj.cx(2, 0); maj.ccx(0, 1, 2);
maj = maj.to_gate();
```

```python
uma = QuantumCircuit(3, name = "UMA");
uma.ccx(0, 1, 2); uma.cx(2, 0); uma.cx(0, 1);
uma = uma.to_gate();
```

```python
qc = QuantumCircuit(10);
qc.x(1); qc.x(3);
qc.x(2); qc.x(4); qc.x(6);
```

```python
qc.append(maj, [0, 1, 2]);
qc.append(maj, [2, 3, 4]);
qc.append(maj, [4, 5, 6]);
qc.append(maj, [6, 7, 8]);
qc.barrier();
qc.cx(8, 9);
qc.append(uma, [6, 7, 8]);
qc.append(uma, [4, 5, 6]);
qc.append(uma, [2, 3, 4]);
qc.append(uma, [0, 1, 2]);
qc.measure_all();
```

```python
qc.draw('mpl')
```

```python
sim = AerSimulator(method = 'statevector')
cc = transpile(qc, sim)
```

```python
res = sim.run(cc).result()
counts = res.get_counts(cc)
counts
```
